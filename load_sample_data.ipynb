```python
import fsspec

fs = fsspec.filesystem(
    "s3",
    client_kwargs={"endpoint_url": "https://s3.echo.stfc.ac.uk"},
    anon=True  # add this if data is public
)
```

```python
import random

files = fs.ls("mast/tokamark/v1")
zarr_files = sorted([f for f in files if f.endswith(".zarr")])

print(f"Total .zarr stores: {len(zarr_files)}")

# show a quick preview
print("\nFirst 5:")
for f in zarr_files[:5]:
    print("  ", f)

print("\nLast 5:")
for f in zarr_files[-5:]:
    print("  ", f)
```

```python
sampled = random.sample(zarr_files, 10)

print("\nRandomly selected shots:")
for f in sampled:
    print("  ", f)
```

```python
for remote_path in sampled:
    shot_name = remote_path.split("/")[-1]
    local_path = f"./data/{shot_name}"
    
    print(f"\nDownloading {shot_name}...")
    
    fs.get(remote_path, local_path, recursive=True)
    
    print(f"Saved to {local_path}")
```
